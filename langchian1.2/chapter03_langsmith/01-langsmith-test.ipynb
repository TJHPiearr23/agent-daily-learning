```python
from langchain_deepseek import ChatDeepSeek
from dotenv import load_dotenv
import os

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_API_BASE = os.getenv("DEEPSEEK_API_BASE")

model = ChatDeepSeek(
    model_name="deepseek-flash",
    api_key=DEEPSEEK_API_KEY,
)

print(model.invoke("你好！"))
```

举例3：

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
from rich import print as rprint

# 1. 从 .env 文件中加载环境变量
load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
# 加了 or 兜底，防止读不到环境变量时变成 None 报错
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL") or "https://api.deepseek.com"

# 打印确认一下是否真的读到了
print(f"API_KEY Loaded: {bool(DEEPSEEK_API_KEY)}")
print(f"BASE_URL Loaded: {DEEPSEEK_BASE_URL}")

# 2. 初始化模型
model = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
    temperature=0.2,
    max_tokens=500,
    # 指定可调整参数
    configurable_fields=("model", "model_provider", "temperature", "max_tokens"),
)

# 3. 准备 config 字典 (修正了括号闭合和逗号问题)
config = {
    "run_name": "joke_generation",  # 在 LangSmith 中这次运行会显示为 joke_generation
    "tags": ["my_tag1", "my_tag2"], # 打上标签便于分类查找
    "metadata": {
        "user_id": "shkstart",      # 记录用户ID
        "session_id": "sess_123"    # 记录会话ID
    },
    "configurable": {
        # 注意：这里改成了 deepseek 的模型，不要把 provider 改成 openai
        "model": "deepseek-v4-pro", 
        "model_provider": "deepseek", # 必须和你的 base_url / api_key 匹配
        "temperature": 0.7, 
        "max_tokens": 1000 
    }
}

# 4. 调用模型并传入 config
response = model.invoke(
    "1 + 2 = ？",
    config=config
)

rprint(response)
```
