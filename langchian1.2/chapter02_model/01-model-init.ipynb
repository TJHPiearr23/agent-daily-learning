# 调用线上的大模型

## 1.调用具体厂商的API

### 1.1 调用Deepseek官网的Deepseek API

调用大模型涉及到三个具体的参数
- base_url
- api_key 
- model_name

举例子：将参数信息放在.env目录中

eg 01：

```python
from langchain_deepseek import ChatDeepSeek
from dotenv import load_dotenv
import os
# 1.加载环境变量, 从.env文件中读取
# override=True表示相关的环境变量以.env文件中的为准, 覆盖系统环境变量
load_dotenv(override=True)

API_KEY = os.getenv("DEEPSEEK_API_KEY")
BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 2.初始化大模型
llm_deepseek = ChatDeepSeek(
    model="deepseek-flash",
    api_key=API_KEY,
    base_url=BASE_URL
) 

# 3.调用大模型
response = llm_deepseek.invoke("你好")
print(response.content)
```

# 2.兼容的写法
使用ChatOpenAI
eg 01:

```python
# 加载配置文件
from dotenv import load_dotenv
load_dotenv(override=True)
from langchain_openai import ChatOpenAI

API_KEY = os.getenv("DEEPSEEK_API_KEY")
BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 获取大模型
model = ChatOpenAI(
    model="deepseek-flash",
    api_key=API_KEY,
    base_url=BASE_URL,
)

response = model.invoke("1+1等于多少？")  
print(response.content)
```

eg 02:调用智谱模型

```python
load_dotenv(override=True)
from langchain_openai import ChatOpenAI
import os

API_KEY = os.getenv("ZHIPUAI_API_KEY")
BASE_URL = os.getenv("ZHIPUAI_BASE_URL")

# 获取大模型
model = ChatOpenAI(
    model="glm-5.3",
    api_key=API_KEY,
    base_url=BASE_URL,
)

model_response = model.invoke("介绍一下langchain")
print(model_response.content)
```
