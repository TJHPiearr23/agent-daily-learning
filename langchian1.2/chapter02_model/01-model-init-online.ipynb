## 1.调用具体厂商的API

### 1.1 调用Deepseek官网的Deepseek API

调用大模型涉及到三个具体的参数
- base_url
- api_key 
- model_name

举例子：将参数信息放在.env目录中

eg 01：

```python
from langchain_deepseek import ChatDeepSeek
from dotenv import load_dotenv
import os
# 1.加载环境变量, 从.env文件中读取
# override=True表示相关的环境变量以.env文件中的为准, 覆盖系统环境变量
load_dotenv(override=True)

API_KEY = os.getenv("DEEPSEEK_API_KEY")
BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 2.初始化大模型
llm_deepseek = ChatDeepSeek(
    model="deepseek-flash",
    api_key=API_KEY,
    base_url=BASE_URL
) 

# 3.调用大模型
response = llm_deepseek.invoke("你好")
print(response.content)
```

# 2.兼容的写法
使用ChatOpenAI
eg 01:

```python
# 加载配置文件
from dotenv import load_dotenv
load_dotenv(override=True)
from langchain_openai import ChatOpenAI

API_KEY = os.getenv("DEEPSEEK_API_KEY")
BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 获取大模型
model = ChatOpenAI(
    model="deepseek-flash",
    api_key=API_KEY,
    base_url=BASE_URL,
)

response = model.invoke("1+1等于多少？")  
print(response.content)
```

# 4. 使用langchain初始化模型
使用init_chat_model函数初始化模型

eg01:

```python
from langchain.chat_models import init_chat_model
import os
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("DEEPSEEK_API_KEY")
BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    model="deepseek-flash",
    model_provider="deepseek",# 指定模型提供者为deepseek
    api_key=API_KEY,
    base_url=BASE_URL
)

model.invoke("你好")
```

或者使用这个写法

```python

model2 = init_chat_model(
    model="deepseek:deepseek-flash",
    api_key=API_KEY,
    base_url=BASE_URL
)

print(model2.invoke("你好").content)
```
